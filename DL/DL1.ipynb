#### 핵심 관계
- 인공지능(AI): 가장 넓은 개념
- 머신러닝(ML): 데이터로부터 패턴을 학습하는 인공지능의 한 분야
- 딥러닝(DL): 여러 층으로 구성된 신경망을 사용하는 머신러닝의 한 분야
- 즉, 인공지능 ⊃ 머신러닝 ⊃ 딥러닝 관계
- 생성형 AI와 LLM도 딥러닝 Transformer를 기반으로 하고 있다.

1. 머신러닝과 딥러닝의 차이
- 머신러닝에서는 KNN, 의사결정나무, 랜덤 포레스트, 로지스틱 회귀 등 사용할 알고리즘을 선택하고 모델을 생성한다.
- 반면 딥러닝에서는 신경망의 층과 뉴런 수, 활성화 함수 등을 구성하여 모델을 설계하고, 학습 과정에서 가중치를 조정한다.

| 구분 | 머신러닝 | 딥러닝 |
|---|---|---|
| 대표적인 모델 | KNN, Random Forest, Logistic Regression | DNN, CNN, RNN, Transformer |
| 모델 구성 | 알고리즘에 맞는 모델 생성 | 신경망의 층과 뉴런 등을 설계 |
| 학습 방법 | 알고리즘에 따라 학습 | 손실 함수와 최적화 알고리즘 등을 이용해 학습 |
| 주요 라이브러리 | `scikit-learn` | `TensorFlow`, `Keras`, `PyTorch` |

2. 머신러닝은 알고리즘을 선택한다
- 예를 들어 KNN을 사용한다면 다음과 같이 모델을 생성한다.

````python
from sklearn.neighbors import KNeighborsClassifier 
# model = KNeighborsClassifier(n_neighbors=5) # KNN 모델 생성 
# model.fit(X_train, y_train) # 모델 학습 
````

````python
from tensorflow.keras import Sequential 
from tensorflow.keras.layers import Dense 
model = Sequential() # 신경망 모델 생성
model.add(Dense(100, activation='relu'))  # 은닉층 추가 
model.add(Dense(3, activation='softmax')) # 출력층 추가 
````

- Sequential : 신경망의 층을 순서대로 연결한다.
- Dense : 이전 층의 출력과 모든 뉴런을 연결하는 완전연결층을 만든다.
- 100 : 해당 층의 뉴런 수를 지정한다.
- relu : 은닉층에서 사용하는 활성화 함수다.
- softmax : 3개 클래스의 예측 확률을 계산한다.
<br>

이처럼 딥러닝은 신경망 구조를 설계하는 과정이 중요하다.
